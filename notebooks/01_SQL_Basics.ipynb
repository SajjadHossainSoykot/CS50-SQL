# 01: SQL Basics (CS50 SQL)
This notebook walks through the basic commands of SQL: `SELECT`, `WHERE`, `LIKE`, `ORDER BY`, and `LIMIT` using Python's `sqlite3` and `pandas` libraries on the `favorites` database.

### Setup: Connect to Database
We establish a connection to our database and check the table structure.

```python
import os
import sqlite3
import pandas as pd

# Locate the database
db_path = os.path.join("..", "databases", "favorites.db")
conn = sqlite3.connect(db_path)

# Verify the schema of favorites table
pd.read_sql_query("PRAGMA table_info(favorites);", conn)
```

### 1. The `SELECT` Clause
**What SQL concept is demonstrated?**
We query specific columns or all columns from the `favorites` table using `SELECT`.

```python
# Querying all columns but limiting rows to 5
query = """
SELECT * 
FROM favorites 
LIMIT 5;
"""
pd.read_sql_query(query, conn)
```

```python
# Selecting specific columns
query = """
SELECT language, problem 
FROM favorites 
LIMIT 5;
"""
pd.read_sql_query(query, conn)
```

### 2. The `WHERE` Clause
**What SQL concept is demonstrated?**
We filter rows where a specific condition is met.

```python
# Selecting only records where the language is Python
query = """
SELECT id, Timestamp, language, problem 
FROM favorites 
WHERE language = 'Python'
LIMIT 5;
"""
pd.read_sql_query(query, conn)
```

### 3. The `LIKE` Wildcard Matching
**What SQL concept is demonstrated?**
Using `LIKE` with wildcards (`%`) to search for string patterns inside text columns.

```python
# Find records where problem contains 'Scratch' anywhere in the name
query = """
SELECT * 
FROM favorites 
WHERE problem LIKE '%Scratch%'
LIMIT 5;
"""
pd.read_sql_query(query, conn)
```

### 4. Sorting with `ORDER BY`
**What SQL concept is demonstrated?**
Sorting result sets alphabetically or numerically in ascending (`ASC`) or descending (`DESC`) order.

```python
# Sort distinct problems alphabetically
query = """
SELECT DISTINCT problem 
FROM favorites 
ORDER BY problem ASC;
"""
pd.read_sql_query(query, conn)
```

### Clean Up
We close the database connection.

```python
conn.close()
print("Connection closed.")
```

### Reflection
**What did I learn?**
- `SELECT` retrieves specific columns of data.
- `WHERE` enables row-level filtering.
- `LIKE` allows fuzzy text matching using `%` as wildcards.
- `ORDER BY` lets us sort output tables cleanly.
- `LIMIT` controls the size of output tables, which is useful when dealing with large datasets.
