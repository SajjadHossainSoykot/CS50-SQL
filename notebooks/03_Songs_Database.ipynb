# 03: Songs Database Analysis (CS50 SQL)
This notebook runs SQL queries on the `songs` database (`songs.db`) containing the top 100 Spotify songs of 2018. It covers selecting, sorting, filtering, aggregations, subqueries, and table joins.

### Setup: Connect to the Database
We establish a connection to our database and check the schemas of the `songs` and `artists` tables.

```python
import os
import sqlite3
import pandas as pd

# Locate the database
db_path = os.path.join("..", "databases", "songs.db")
conn = sqlite3.connect(db_path)

# Verify the tables
pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)
```

### 1. Retrieve all Song Names
**What SQL concept is demonstrated?**
A basic `SELECT` query returning the names of all songs.

```python
query1 = """
SELECT name
FROM songs;
"""
pd.read_sql_query(query1, conn).head(10)
```

### 2. Sort Songs by Tempo
**What SQL concept is demonstrated?**
Sorting output rows using the `ORDER BY` clause (defaulting to ascending order).

```python
query2 = """
SELECT name
FROM songs
ORDER BY tempo;
"""
pd.read_sql_query(query2, conn).head(10)
```

### 3. Top 5 Longest Songs
**What SQL concept is demonstrated?**
Sorting rows in descending order (`ORDER BY ... DESC`) combined with `LIMIT` to get the top N records.

```python
query3 = """
SELECT name
FROM songs
ORDER BY duration_ms DESC
LIMIT 5;
"""
pd.read_sql_query(query3, conn)
```

### 4. High Danceability, Energy, and Valence Songs
**What SQL concept is demonstrated?**
Filtering using the `WHERE` clause combined with multiple conditions using the `AND` operator.

```python
query4 = """
SELECT name
FROM songs
WHERE danceability > 0.75
AND energy > 0.75
AND valence > 0.75;
"""
pd.read_sql_query(query4, conn)
```

### 5. Average Energy of All Songs
**What SQL concept is demonstrated?**
Using the `AVG()` aggregate function on numerical columns.

```python
query5 = """
SELECT AVG(energy)
FROM songs;
"""
pd.read_sql_query(query5, conn)
```

### 6. Songs by Post Malone (Subquery)
**What SQL concept is demonstrated?**
Using a subquery (nested query) to find the `artist_id` associated with a specific artist name.

```python
query6 = """
SELECT name
FROM songs
WHERE artist_id =
(
    SELECT id
    FROM artists
    WHERE name = 'Post Malone'
);
"""
pd.read_sql_query(query6, conn)
```

### 7. Average Energy of Songs by Drake (Table Join)
**What SQL concept is demonstrated?**
Combining records from two tables (`songs` and `artists`) using an `INNER JOIN` (or simple `JOIN`).

```python
query7 = """
SELECT AVG(energy)
FROM songs
JOIN artists ON songs.artist_id = artists.id
WHERE artists.name = 'Drake';
"""
pd.read_sql_query(query7, conn)
```

### 8. Featured Songs
**What SQL concept is demonstrated?**
Fuzzy string pattern matching using `LIKE` with wildcards (`%`).

```python
query8 = """
SELECT name
FROM songs
WHERE name LIKE '%feat.%';
"""
pd.read_sql_query(query8, conn)
```
