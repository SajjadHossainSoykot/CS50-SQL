## 📊 Section 2: Aggregation & Counting
How to aggregate and count occurrences of each language option.

### 5. `favorites4.py` - Counting with Standard Variables
Counts popularity using manual conditional checks (`if/elif`).

```python
import csv

# Open CSV file
with open("favorites.csv", "r") as file:
    # Create DictReader
    reader = csv.DictReader(file)
    scratch, c, python = 0, 0, 0
    # Iterate over CSV file, counting favorites
    for row in reader:
        favorite = row["language"]
        if favorite == "Scratch":
            scratch += 1
        elif favorite == "C":
            c += 1
        elif favorite == "Python":
            python += 1

# Print counts
print(f"Scratch: {scratch}")
print(f"C: {c}")
print(f"Python: {python}")
```

### 6. `favorites5.py` - Counting with Dictionaries
Uses a standard Python dictionary `counts` to count any programming language automatically.

```python
import csv

# Open CSV file
with open("favorites.csv", "r") as file:
    # Create DictReader
    reader = csv.DictReader(file)
    counts = {}
    # Iterate over CSV file, counting favorites
    for row in reader:
        favorite = row["language"]
        if favorite in counts:
            counts[favorite] += 1
        else:
            counts[favorite] = 1

# Print counts
for favorite in counts:
    print(f"{favorite}: {counts[favorite]}")
```

### 7. `favorites6.py` - Counting using Try/Except
Uses exception handling (`try/except KeyError`) to initialize new keys in the dictionary.

```python
import csv

# Open CSV file
with open("favorites.csv", "r") as file:
    # Create DictReader
    reader = csv.DictReader(file)
    counts = {}
    # Iterate over CSV file, counting favorites
    for row in reader:
        favorite = row["language"]
        try:
            counts[favorite] += 1
        except KeyError:
            counts[favorite] = 1

# Print counts
for favorite in counts:
    print(f"{favorite}: {counts[favorite]}")
```

## 🔀 Section 3: Sorting Aggregated Data

### 8. `favorites7.py` - Alphabetical Sorting
Sorts the counted dictionary keys alphabetically.

```python
import csv

# Open CSV file
with open("favorites.csv", "r") as file:
    # Create DictReader
    reader = csv.DictReader(file)
    counts = {}
    # Iterate over CSV file, counting favorites
    for row in reader:
        favorite = row["language"]
        if favorite in counts:
            counts[favorite] += 1
        else:
            counts[favorite] = 1

# Print counts sorted by language name
for favorite in sorted(counts):
    print(f"{favorite}: {counts[favorite]}")
```

### 9. `favorites8.py` - Sorting by Popularity
Sorts the counted dictionary keys by their count values in descending order using `counts.get`.

```python
import csv

# Open CSV file
with open("favorites.csv", "r") as file:
    # Create DictReader
    reader = csv.DictReader(file)
    counts = {}
    # Iterate over CSV file, counting favorites
    for row in reader:
        favorite = row["language"]
        if favorite in counts:
            counts[favorite] += 1
        else:
            counts[favorite] = 1

# Print counts sorted by popularity (value)
for favorite in sorted(counts, key=counts.get, reverse=True):
    print(f"{favorite}: {counts[favorite]}")
```

## 🗄️ Section 4: Relational Databases & SQL
Now we transition to storing the data in a relational database (`favorites.db`).

### 10. `csv_to_db.py` - Populating the Database
Reads `favorites.csv` and inserts the rows into a new SQLite table called `favorites`.

```python
import os
import subprocess

# Reset/remove the existing database file to start clean
if os.path.exists("favorites.db"):
    os.remove("favorites.db")

# Define SQLite CLI commands to import the CSV while keeping the autoincrement ID
commands = """
.mode csv
.import favorites.csv temp_favorites

CREATE TABLE favorites (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    Timestamp TEXT,
    language TEXT,
    problem TEXT
);

INSERT INTO favorites (Timestamp, language, problem)
SELECT Timestamp, language, problem FROM temp_favorites;

DROP TABLE temp_favorites;
"""

# Run the SQLite CLI commands
subprocess.run(["sqlite3", "favorites.db"], input=commands, text=True)

print("Database favorites.db successfully populated using SQLite CLI!")
```

### 11. `favorites9.py` - Querying & Aggregating via SQL
Using SQL queries to group, count, and sort the favorites table.

```python
from cs50 import SQL

# Open database
db = SQL("sqlite:///favorites.db")

# Search for title
rows = db.execute("SELECT language, COUNT(*) AS n FROM favorites GROUP BY language ORDER BY n DESC")

# Print popularity
for row in rows:
    print(row["language"], row["n"])
```

### 12. `favorites10.py` - Querying with Parameters
Queries popularity for a specific problem inputted by the user using placeholders (`?`) to prevent SQL injection.

```python
from cs50 import SQL

# Open database
db = SQL("sqlite:///favorites.db")

# Prompt user for favorite problem
favorite = input("Favorite (e.g., Readability, Mario, Filter, Credit, Cash): ")

# Search for title
rows = db.execute("SELECT COUNT(*) AS n FROM favorites WHERE problem = ?", favorite)

# Get first (and only) row
row = rows[0]

# Print popularity
print(f"Popularity of {favorite}: {row['n']}")
```
